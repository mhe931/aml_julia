```python
import os
import sys
import time
import pickle
import numpy as np
import pandas as pd
import warnings

# Attempt to import GPU-accelerated libraries
try:
    import cuml
    from cuml.cluster import KMeans as cuKMeans
    from cuml.decomposition import PCA as cuPCA
    from cuml.metrics.cluster import silhouette_score as cu_silhouette_score
    from cuml.preprocessing import StandardScaler as cuStandardScaler
    # Check for IsolationForest in cuml
    try:
        from cuml.ensemble import IsolationForest as cuIsolationForest
    except ImportError:
        from cuml.solvers import CD as cuIsolationForest 
        pass
    import cupy as cp
    print("RAPIDS cuML and CuPy imported successfully.")
except ImportError as e:
    print(f"WARNING: RAPIDS cuML/CuPy not found. This script requires a GPU environment. Error: {e}")

try:
    import torch
    import torch.nn as nn
    import torch.optim as optim
    from torch.utils.data import DataLoader, TensorDataset
    print("PyTorch imported successfully.")
except ImportError:
    print("WARNING: PyTorch not found. Autoencoder section will fail.")

from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import davies_bouldin_score
```

```python
# Configuration
DATA_FILE = 'master_data.csv'
OUTPUT_FILE = 'analysis_summary.txt'
MODEL_DIR = 'models_gpu'
if not os.path.exists(MODEL_DIR):
    os.makedirs(MODEL_DIR)
```

## Data Preprocessing & Feature Engineering
- Load `master_data.csv`
- Drop low-variance features
- Impute missing values
- Encode categorical features
- Scale data (GPU)
- Perform PCA (GPU)

```python
def load_and_preprocess_data(filepath):
    print("Loading data...")
    if not os.path.exists(filepath):
        print(f"Error: {filepath} not found.")
        return None, None, None, None
    
    df = pd.read_csv(filepath)
    print(f"Data loaded: {df.shape}")

    # 1. Drop low variance features
    nunique = df.nunique()
    cols_to_drop = nunique[nunique <= 1].index
    df = df.drop(cols_to_drop, axis=1)
    print(f"Dropped {len(cols_to_drop)} low-variance columns.")

    # 2. Imputation
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    categorical_cols = df.select_dtypes(exclude=[np.number]).columns

    for col in numeric_cols:
        df[col] = df[col].fillna(df[col].mean())
    
    for col in categorical_cols:
        if len(df[col].mode()) > 0:
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna("Unknown")

    # 3. Encoding
    print("Encoding categorical features...")
    df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
    
    # Convert to float32 for GPU
    data_np = df_encoded.values.astype(np.float32)
    
    # 4. Scaling (GPU)
    print("Scaling data (GPU)...")
    scaler = cuStandardScaler()
    data_gpu = cp.array(data_np)
    data_scaled = scaler.fit_transform(data_gpu)
    
    # 5. PCA (GPU)
    print("Running PCA (GPU)...")
    pca = cuPCA(n_components=0.95)
    data_pca = pca.fit_transform(data_scaled)
    print(f"Data shape after PCA: {data_pca.shape}")
    
    return data_pca, df_encoded, scaler, pca
```

```python
# Execute Loading
data_pca, df_encoded, scaler, pca_model = load_and_preprocess_data(DATA_FILE)
```

## Model 1: K-Means Optimization
- Test range of K (5-20)
- Use Silhouette Score to find optimal K

```python
def train_kmeans_gpu(data, k_range):
    print("Training K-Means (GPU)...")
    best_score = -1
    best_k = -1
    best_model = None
    
    results = {}

    for k in k_range:
        kmeans = cuKMeans(n_clusters=k, random_state=42)
        kmeans.fit(data)
        labels = kmeans.labels_
        
        # Silhouette Score
        score = cu_silhouette_score(data, labels)
        results[k] = score
        print(f"K={k}, Silhouette={score:.4f}")
        
        if score > best_score:
            best_score = score
            best_k = k
            best_model = kmeans
            
    return best_model, best_k, best_score, results
```

```python
if data_pca is not None:
    kmeans_model, best_k_kmeans, best_score_kmeans, _ = train_kmeans_gpu(data_pca, range(5, 21))
    
    # Save Model
    with open(os.path.join(MODEL_DIR, 'kmeans_model.pkl'), 'wb') as f:
        pickle.dump(kmeans_model, f)
```

## Model 2: X-Means (Simulated)
- Simulate X-Means using BIC score to select optimal K

```python
def train_xmeans_simulated(data, max_k=20):
    print("Training X-Means (Simulated via BIC)...")
    best_bic = np.inf
    best_k = -1
    best_model = None
    
    n_samples, _ = data.shape
    
    for k in range(2, max_k + 1):
        kmeans = cuKMeans(n_clusters=k, random_state=42)
        kmeans.fit(data)
        
        # Calculate BIC (Heuristic)
        inertia = kmeans.inertia_
        bic = n_samples * np.log(inertia / n_samples) + k * np.log(n_samples)
        
        if bic < best_bic:
            best_bic = bic
            best_k = k
            best_model = kmeans
            
    return best_model, best_k
```

```python
if data_pca is not None:
    xmeans_model, best_k_xmeans = train_xmeans_simulated(data_pca)
    
    with open(os.path.join(MODEL_DIR, 'xmeans_model.pkl'), 'wb') as f:
        pickle.dump(xmeans_model, f)
```

## Model 3: Autoencoder + K-Means
- Train a Shallow Autoencoder (PyTorch)
- Cluster on the Latent Space

```python
class Autoencoder(nn.Module):
    def __init__(self, input_dim, encoding_dim):
        super(Autoencoder, self).__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, encoding_dim),
            nn.ReLU() 
        )
        self.decoder = nn.Sequential(
            nn.Linear(encoding_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, input_dim),
            nn.Sigmoid() 
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return encoded, decoded

def train_autoencoder_clustering(data_np, encoding_dim=10, epochs=20):
    print("Training Autoencoder...")
    input_dim = data_np.shape[1]
    
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Autoencoder running on: {device}")
    
    model = Autoencoder(input_dim, encoding_dim).to(device)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    
    if hasattr(data_np, 'get'):
        data_np = data_np.get()
        
    tensor_x = torch.Tensor(data_np).to(device)
    dataset = TensorDataset(tensor_x, tensor_x)
    dataloader = DataLoader(dataset, batch_size=256, shuffle=True)
    
    model.train()
    for epoch in range(epochs):
        total_loss = 0
        for batch_features, _ in dataloader:
            optimizer.zero_grad()
            encoded, decoded = model(batch_features)
            loss = criterion(decoded, batch_features)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        print(f"Epoch {epoch+1}/{epochs}, Loss: {total_loss/len(dataloader):.4f}")
        
    model.eval()
    with torch.no_grad():
        latent_features, _ = model(tensor_x)
    
    latent_np = latent_features.cpu().numpy()
    latent_gpu = cp.array(latent_np)
    
    print("Clustering on Latent Space...")
    kmeans_ae = cuKMeans(n_clusters=5, random_state=42)
    kmeans_ae.fit(latent_gpu)
    
    return model, kmeans_ae, latent_gpu
```

```python
if data_pca is not None:
    ae_model, ae_kmeans, latent_data = train_autoencoder_clustering(data_pca)
    
    torch.save(ae_model.state_dict(), os.path.join(MODEL_DIR, 'autoencoder_weights.pth'))
    with open(os.path.join(MODEL_DIR, 'ae_kmeans_model.pkl'), 'wb') as f:
        pickle.dump(ae_kmeans, f)
```

## Result Aggregation & Analysis
- Calculate Outliers
- Calculate Davies-Bouldin Index
- Identify Top Features
- Generate Summary Report

```python
def calculate_outliers(data, labels, centroids):
    point_centroids = centroids[labels]
    diff = data - point_centroids
    dist_sq = cp.sum(diff**2, axis=1)
    distances = cp.sqrt(dist_sq)
    threshold = cp.percentile(distances, 95)
    outliers = distances > threshold
    return outliers, threshold

def get_top_features(numeric_df, labels, top_n=3):
    if hasattr(labels, 'get'):
        labels = labels.get()
        
    X = numeric_df.copy()
    y = pd.Series(labels, index=X.index)
    
    clf = DecisionTreeClassifier(max_depth=5)
    clf.fit(X, y)
    importances = clf.feature_importances_
    indices = np.argsort(importances)[::-1]
    
    top_features = [X.columns[i] for i in indices[:top_n]]
    return top_features
```

```python
if data_pca is not None:
    results_summary = []
    
    # --- K-Means Analysis ---
    outliers_km, thresh_km = calculate_outliers(data_pca, kmeans_model.labels_, kmeans_model.cluster_centers_)
    num_outliers_km = cp.sum(outliers_km).item()
    pct_outliers_km = (num_outliers_km / len(data_pca)) * 100
    
    data_np = cp.asnumpy(data_pca)
    labels_km_np = cp.asnumpy(kmeans_model.labels_)
    db_score_km = davies_bouldin_score(data_np, labels_km_np)
    top_feats_km = get_top_features(df_encoded, labels_km_np)
    
    results_summary.append({
        "Model": "K-Means (Optimized)",
        "Optimal_K": best_k_kmeans,
        "Silhouette": best_score_kmeans,
        "Davies_Bouldin": db_score_km,
        "Outliers_Count": num_outliers_km,
        "Outliers_Pct": pct_outliers_km,
        "Top_Features": top_feats_km
    })
    
    # --- X-Means Analysis ---
    labels_xm = xmeans_model.labels_
    score_xm = cu_silhouette_score(data_pca, labels_xm)
    labels_xm_np = cp.asnumpy(labels_xm)
    db_score_xm = davies_bouldin_score(data_np, labels_xm_np)
    
    outliers_xm, thresh_xm = calculate_outliers(data_pca, labels_xm, xmeans_model.cluster_centers_)
    num_outliers_xm = cp.sum(outliers_xm).item()
    pct_outliers_xm = (num_outliers_xm / len(data_pca)) * 100
    top_feats_xm = get_top_features(df_encoded, labels_xm_np)
    
    results_summary.append({
        "Model": "X-Means (Simulated)",
        "Optimal_K": best_k_xmeans,
        "Silhouette": score_xm,
        "Davies_Bouldin": db_score_xm,
        "Outliers_Count": num_outliers_xm,
        "Outliers_Pct": pct_outliers_xm,
        "Top_Features": top_feats_xm
    })
    
    # --- Autoencoder Analysis ---
    labels_ae = ae_kmeans.labels_
    score_ae = cu_silhouette_score(latent_data, labels_ae)
    labels_ae_np = cp.asnumpy(labels_ae)
    latent_np = cp.asnumpy(latent_data)
    db_score_ae = davies_bouldin_score(latent_np, labels_ae_np)
    
    outliers_ae, thresh_ae = calculate_outliers(latent_data, labels_ae, ae_kmeans.cluster_centers_)
    num_outliers_ae = cp.sum(outliers_ae).item()
    pct_outliers_ae = (num_outliers_ae / len(data_pca)) * 100
    top_feats_ae = get_top_features(df_encoded, labels_ae_np)
    
    results_summary.append({
        "Model": "Autoencoder + K-Means",
        "Optimal_K": ae_kmeans.n_clusters,
        "Silhouette": score_ae,
        "Davies_Bouldin": db_score_ae,
        "Outliers_Count": num_outliers_ae,
        "Outliers_Pct": pct_outliers_ae,
        "Top_Features": top_feats_ae
    })
    
    # Write Summary
    print(f"\nWriting results to {OUTPUT_FILE}...")
    with open(OUTPUT_FILE, 'w') as f:
        f.write("Advanced Unsupervised Clustering Analysis Summary\n")
        f.write("===============================================\n\n")
        
        for res in results_summary:
            f.write(f"Model: {res['Model']}\n")
            f.write(f"-----------------------------------------------\n")
            f.write(f"Optimal Clusters (K): {res['Optimal_K']}\n")
            f.write(f"Silhouette Score:     {res['Silhouette']:.4f}\n")
            f.write(f"Davies-Bouldin Index: {res['Davies_Bouldin']:.4f}\n")
            f.write(f"Top 3 Features:       {', '.join(res['Top_Features'])}\n")
            f.write(f"Outliers Detected:    {res['Outliers_Count']} ({res['Outliers_Pct']:.2f}%)\n")
            f.write("\n")
            
    print("Analysis Complete.")
    # Print content of summary for verification
    with open(OUTPUT_FILE, 'r') as f:
        print(f.read())
```
